```python
import google.auth

credentials, project = google.auth.default()

print("Active project:", project)
print("Credential type:", type(credentials).__name__)
print("Service account:", getattr(credentials, "service_account_email", "Not directly available"))
```

```python
import subprocess

result = subprocess.run(
    ["gcloud", "auth", "list", "--filter=status:ACTIVE",
     "--format=value(account)"],
    capture_output=True,
    text=True
)

print("Active gcloud account:")
print(result.stdout if result.stdout.strip() else result.stderr)
```

```python
from google.cloud import storage

bucket_name = "breast-cancer-model"

try:
    client = storage.Client()
    bucket = client.get_bucket(bucket_name)

    print("SUCCESS: Cloud Storage access verified!")
    print("Bucket:", bucket.name)
    print("Location:", bucket.location)

    blob = bucket.blob("breast_cancer_model.pkl")
    print("Model exists:", blob.exists())

except Exception as e:
    print("Cloud Storage access failed:")
    print(type(e).__name__, str(e))
    
```

```python
from google.cloud import secretmanager

project_id = "cloud-computing-project-510701"
secret_id = "hospital-a-api-key"

client = secretmanager.SecretManagerServiceClient()

secret_name = (
    f"projects/{project_id}/secrets/"
    f"{secret_id}/versions/latest"
)

try:
    response = client.access_secret_version(
        request={"name": secret_name}
    )

    print("SUCCESS: Secret Manager access verified!")
    print("Secret:", secret_id)
    print("Secret version:", response.name.split("/")[-1])
    print("Secret value: NOT DISPLAYED")

except Exception as e:
    print("Secret Manager access failed:")
    print(type(e).__name__, str(e))
```
