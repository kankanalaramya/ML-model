```python
import subprocess

command = [
    "gcloud", "artifacts", "repositories", "describe",
    "ramya-prediction-api",
    "--location=northamerica-northeast2",
    "--project=cloud-computing-project-510701"
]

result = subprocess.run(
    command,
    capture_output=True,
    text=True
)

print("Return code:", result.returncode)
print(result.stdout)

if result.stderr:
    print("Error:", result.stderr)
```

```python
import subprocess
import os

# Check Docker version
result = subprocess.run(
    ["docker", "--version"],
    capture_output=True,
    text=True
)

print("DOCKER VERSION:")
print(result.stdout or result.stderr)

# Check Docker availability
result = subprocess.run(
    ["docker", "info"],
    capture_output=True,
    text=True
)

print("\nDOCKER STATUS:")
print(result.stdout[:1500] if result.returncode == 0 else result.stderr)

# Check project files
print("\nPROJECT FILES:")
for filename in ["app.py", "Dockerfile", "requirements.txt"]:
    print(filename, ":", "FOUND" if os.path.exists(filename) else "MISSING")
```
